```python
%%bash
# stopping an OSD
function stop_osd() {
  local prim_osd=$1
  procid=$(cat out/osd.$prim_osd.pid)
  echo "Stopping OSD $prim_osd with PID $procid"
  echo kill -TERM $procid
}

prim_osd=$(bin/ceph --format=json-pretty osd map ecpool dummy | jq -r '.acting_primary')
stop_osd $prim_osd
sleep 2

bin/ceph-objectstore-tool --data-path ./asok/ --pgid 11.14 --op list

```
